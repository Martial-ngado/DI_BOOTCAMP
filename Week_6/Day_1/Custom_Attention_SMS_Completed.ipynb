# Part 4: Custom Attention Implementation
Build the simple attention layer, classifier, and data pipeline for the scratch model.


> **Learning point**
> Scaling the dot products by $1/\sqrt{d_k}$ keeps gradients stable and prevents the softmax from collapsing when embeddings grow. This opeeration is crucial for training deep attention models.

```python
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import torch.nn as nn


class Attention(nn.Module):
    def __init__(self, embed_dim):
        super().__init__()
        self.scale = embed_dim ** -0.5

    def forward(self, query, key, value, mask=None):
        scores = torch.matmul(
            query, key.transpose(-2, -1)
        ) * self.scale
        if mask is not None:
            scores = scores.masked_fill(mask == 0, float('-inf'))
        attn = F.softmax(scores, dim=-1)
        return torch.matmul(attn, value), attn


class SimpleAttentionClassifier(nn.Module):
    def __init__(self, vocab_size, embed_dim, num_classes):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.attn = Attention(embed_dim)
        self.fc = nn.Linear(embed_dim, num_classes)

    def forward(self, x):
        embed = self.embedding(x)
        attn_output, _ = self.attn(embed, embed, embed)
        pooled = attn_output.mean(dim=1)
        return self.fc(pooled)
```

# Part 5: Metrics & Evaluation
Load accuracy, precision, recall, and F1 from `evaluate`, then implement the shared `compute_metrics` helper.


```python
import evaluate
import numpy as np

accuracy = evaluate.load('accuracy')
precision = evaluate.load('precision')
recall = evaluate.load('recall')
f1 = evaluate.load('f1')


def compute_metrics(pred):
    logits, labels = pred
    preds = np.argmax(logits, axis=-1)
    return {
        'accuracy': accuracy.compute(predictions=preds, references=labels)['accuracy'],
        'precision': precision.compute(predictions=preds, references=labels)['precision'],
        'recall': recall.compute(predictions=preds, references=labels)['recall'],
        'f1': f1.compute(predictions=preds, references=labels)['f1'],
    }
```

### 1. What are the roles of query, key, and value in the attention mechanism?

The **query** represents what the current token is *looking for* — it encodes the token's information need. The **keys** represent what each token in the sequence *offers* — they are matched against the query to compute relevance scores via dot product. The **values** carry the actual content that gets aggregated: once the softmax over scores decides how much each position contributes, the output is a weighted sum of the values. In self-attention all three come from the same sequence, so every token can dynamically draw on any other position's content based on learned compatibility.


### 2. Why do we use a scaling factor in the dot-product attention?

When the embedding dimension $d_k$ is large, the raw dot products $q \cdot k^T$ grow in magnitude (variance scales with $d_k$). This pushes the softmax into regions where its gradient is nearly zero — the distribution becomes extremely peaked and updates vanish. Dividing by $\sqrt{d_k}$ brings the scores back to unit variance regardless of model size, keeping the softmax in a well-behaved regime and making training stable.


### 3. How does self-attention differ from traditional sequence models like RNNs?

RNNs process tokens **sequentially** — hidden state is passed step by step, so long-range dependencies must survive many multiplicative operations and are prone to vanishing gradients. Self-attention is **fully parallel**: every token attends to every other token in a single matrix multiplication, giving O(1) path length between any two positions. This makes long-range dependency capture much more reliable and enables efficient GPU utilization. The trade-off is quadratic memory/compute with sequence length, whereas RNNs scale linearly in length but are harder to parallelize.


### 4. Performance analysis

**GPT-2 is expected to outperform the custom model significantly.** GPT-2 was pre-trained on hundreds of gigabytes of text, giving it rich token representations from day one; its classification head only needs to learn a linear mapping on top of already-meaningful features. The custom model, by contrast, learns embeddings from scratch on just 4,000 examples, which is far too few to build robust representations — attention weights will be noisy and the embeddings underfit.

**Trade-offs:** GPT-2 is large (~117 M parameters), slow to fine-tune without a GPU, and requires more memory. The custom attention classifier is tiny and fast but needs more data or regularization to generalize. **To improve the custom model** one could: (1) initialize embeddings from a pre-trained source (e.g., GloVe or the GPT-2 embedding matrix), (2) add multiple attention heads and stacked layers, or (3) train for several epochs instead of one pass to let gradients converge.

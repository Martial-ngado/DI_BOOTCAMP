```python
import numpy as np
from scipy import stats

print("\n" + "=" * 70)
print("EXERCISE 2: Descriptive Statistics")
print("=" * 70)

data = [12, 15, 13, 12, 18, 20, 22, 21]
print(f"Sample Data: {data}")

# Calculate mean using SciPy
mean = np.mean(data)
print(f"Mean: {mean}")

# Calculate median
median = np.median(data)
print(f"Median: {median}")

# Calculate variance
variance = np.var(data, ddof=1)  # Using sample variance (ddof=1)
print(f"Variance (sample): {variance:.4f}")

# Calculate standard deviation
std_dev = np.std(data, ddof=1)  # Using sample standard deviation
print(f"Standard Deviation (sample): {std_dev:.4f}")

# Using SciPy stats functions
print("\nUsing SciPy stats functions:")
print(f"describe(): {stats.describe(data)}")
```

```python
import matplotlib.pyplot as plt
from scipy.stats import norm

print("\n" + "=" * 70)
print("EXERCISE 3: Understanding Distributions")
print("=" * 70)

# Generate a normal distribution with mean 50 and std 10
mean = 50
std = 10
x = np.linspace(mean - 4 * std, mean + 4 * std, 1000)
y = norm.pdf(x, mean, std)

# Plot the distribution
plt.figure(figsize=(10, 6))
plt.plot(x, y, 'b-', linewidth=2, label=f'Normal Distribution (μ={mean}, σ={std})')
plt.fill_between(x, y, alpha=0.3)
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.title('Normal Distribution with Mean=50 and Standard Deviation=10')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

# Generate random samples from the distribution
samples = np.random.normal(mean, std, 1000)
print(f"\nGenerated 1000 random samples from N({mean}, {std}²)")
print(f"Sample mean: {np.mean(samples):.2f}")
print(f"Sample std: {np.std(samples):.2f}")

# Plot histogram with PDF overlay
plt.figure(figsize=(10, 6))
plt.hist(samples, bins=50, density=True, alpha=0.7, color='skyblue', label='Histogram of samples')
plt.plot(x, y, 'r-', linewidth=2, label='PDF')
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Histogram of Normal Distribution Samples with PDF Overlay')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
from scipy.stats import ttest_ind

print("\n" + "=" * 70)
print("EXERCISE 4: T-Test Application")
print("=" * 70)

# Generate random data
np.random.seed(42)
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

print(f"Data1 - Mean: {np.mean(data1):.2f}, Std: {np.std(data1):.2f}")
print(f"Data2 - Mean: {np.mean(data2):.2f}, Std: {np.std(data2):.2f}")

# Perform independent samples t-test
t_statistic, p_value = ttest_ind(data1, data2)
print(f"\nIndependent Samples T-Test Results:")
print(f"T-Statistic: {t_statistic:.4f}")
print(f"P-Value: {p_value:.6f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print(f"\nConclusion: The p-value ({p_value:.6f}) is less than α ({alpha}).")
    print("We reject the null hypothesis. The two groups have significantly different means.")
else:
    print(f"\nConclusion: The p-value ({p_value:.6f}) is greater than or equal to α ({alpha}).")
    print("We fail to reject the null hypothesis. There is no significant difference between the groups.")

# Visualize the distributions
plt.figure(figsize=(10, 6))
plt.hist(data1, bins=30, alpha=0.6, label=f'Data1 (μ≈{np.mean(data1):.1f})', color='skyblue')
plt.hist(data2, bins=30, alpha=0.6, label=f'Data2 (μ≈{np.mean(data2):.1f})', color='salmon')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Distribution Comparison for T-Test')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
from scipy.stats import linregress

print("\n" + "=" * 70)
print("EXERCISE 5: Linear Regression Analysis")
print("=" * 70)

# Housing data
house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])

print(f"House sizes (m²): {house_sizes}")
print(f"House prices: {house_prices}")

# Perform linear regression
slope, intercept, r_value, p_value, std_err = linregress(house_sizes, house_prices)

print(f"\nLinear Regression Results:")
print(f"Slope: {slope:.2f} (price per m²)")
print(f"Intercept: {intercept:.2f}")
print(f"R-squared: {r_value**2:.4f}")
print(f"P-value: {p_value:.6f}")
print(f"Standard Error: {std_err:.2f}")

# Interpret the slope
print(f"\nInterpretation of Slope:")
print(f"For every additional square meter, the house price increases by ${slope:.2f}")

# Predict the price for a house of 90 m²
size_to_predict = 90
predicted_price = slope * size_to_predict + intercept
print(f"\nPrediction:")
print(f"Predicted price for a {size_to_predict} m² house: ${predicted_price:.2f}")

# Plot the regression line
plt.figure(figsize=(10, 6))
plt.scatter(house_sizes, house_prices, color='blue', s=100, label='Actual data points')
x_line = np.linspace(house_sizes.min() - 10, house_sizes.max() + 10, 100)
y_line = slope * x_line + intercept
plt.plot(x_line, y_line, color='red', linewidth=2, label=f'Regression line: y={slope:.2f}x+{intercept:.2f}')
plt.scatter([size_to_predict], [predicted_price], color='green', s=100, marker='*', label=f'Prediction for {size_to_predict} m²')
plt.xlabel('House Size (m²)')
plt.ylabel('House Price')
plt.title('Linear Regression: House Size vs Price')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
from scipy.stats import f_oneway

print("\n" + "=" * 70)
print("EXERCISE 6: Understanding ANOVA")
print("=" * 70)

# Fertilizer data
fertilizer_1 = np.array([5, 6, 7, 6, 5])
fertilizer_2 = np.array([7, 8, 7, 9, 8])
fertilizer_3 = np.array([4, 5, 4, 3, 4])

print(f"Fertilizer 1 growth (cm): {fertilizer_1}")
print(f"Fertilizer 2 growth (cm): {fertilizer_2}")
print(f"Fertilizer 3 growth (cm): {fertilizer_3}")

print(f"\nDescriptive Statistics:")
print(f"Fertilizer 1 - Mean: {np.mean(fertilizer_1):.2f}, Std: {np.std(fertilizer_1):.2f}")
print(f"Fertilizer 2 - Mean: {np.mean(fertilizer_2):.2f}, Std: {np.std(fertilizer_2):.2f}")
print(f"Fertilizer 3 - Mean: {np.mean(fertilizer_3):.2f}, Std: {np.std(fertilizer_3):.2f}")

# Perform ANOVA
f_value, p_value = f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

print(f"\nANOVA Test Results:")
print(f"F-Value: {f_value:.4f}")
print(f"P-Value: {p_value:.6f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print(f"\nConclusion: The p-value ({p_value:.6f}) is less than α ({alpha}).")
    print("We reject the null hypothesis. The fertilizers have significantly different effects on plant growth.")
else:
    print(f"\nConclusion: The p-value ({p_value:.6f}) is greater than or equal to α ({alpha}).")
    print("We fail to reject the null hypothesis. There is no significant difference in the effects of the fertilizers.")

print(f"\nIf p-value > 0.05:")
print("- We would fail to reject the null hypothesis")
print("- The differences in plant growth would not be statistically significant")
print("- The fertilizers would be considered equally effective")

# Visualize the data
plt.figure(figsize=(10, 6))
data = [fertilizer_1, fertilizer_2, fertilizer_3]
labels = ['Fertilizer 1', 'Fertilizer 2', 'Fertilizer 3']
box_plot = plt.boxplot(data, labels=labels, patch_artist=True)
for patch, color in zip(box_plot['boxes'], ['lightblue', 'lightgreen', 'lightyellow']):
    patch.set_facecolor(color)
plt.ylabel('Plant Growth (cm)')
plt.title('ANOVA: Plant Growth by Fertilizer Type')
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```

```python
from scipy.stats import binom

print("\n" + "=" * 70)
print("EXERCISE 7: Probability Distributions (Optional)")
print("=" * 70)

# Binomial distribution: probability of getting exactly k successes in n trials with probability p
n = 10  # Number of coin flips
p = 0.5  # Probability of heads
k_values = np.arange(0, n + 1)

# Calculate probabilities
probabilities = binom.pmf(k_values, n, p)

print(f"Binomial Distribution: n={n} trials, p={p} (fair coin)")
print(f"\nProbability of getting exactly k heads:")
for k, prob in zip(k_values, probabilities):
    print(f"P(X = {k:2d}) = {prob:.6f}")

# Probability of exactly 5 heads
prob_5_heads = binom.pmf(5, n, p)
print(f"\nProbability of exactly 5 heads in 10 flips: {prob_5_heads:.6f}")

# Cumulative probability: P(X <= k)
print(f"\nCumulative probabilities:")
for k in [3, 5, 7]:
    cumulative_prob = binom.cdf(k, n, p)
    print(f"P(X ≤ {k}) = {cumulative_prob:.6f}")

# Plot binomial distribution
plt.figure(figsize=(10, 6))
plt.bar(k_values, probabilities, color='skyblue', edgecolor='navy', alpha=0.7)
plt.axvline(x=5, color='red', linestyle='--', linewidth=2, label='P(X=5)')
plt.xlabel('Number of Heads')
plt.ylabel('Probability')
plt.title(f'Binomial Distribution (n={n}, p={p})')
plt.xticks(k_values)
plt.legend()
plt.grid(True, alpha=0.3, axis='y')
plt.show()

# Additional example: Probability of at least 7 heads
prob_at_least_7 = 1 - binom.cdf(6, n, p)
print(f"\nProbability of at least 7 heads: {prob_at_least_7:.6f}")
```

```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr

print("\n" + "=" * 70)
print("EXERCISE 8: Correlation Coefficients (Optional)")
print("=" * 70)

# Create a sample dataset
data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})

print("Dataset:")
print(data)

# Calculate Pearson correlation
pearson_corr, pearson_pvalue = pearsonr(data['age'], data['income'])
print(f"\nPearson Correlation Coefficient: {pearson_corr:.6f}")
print(f"P-Value: {pearson_pvalue:.6f}")
print(f"Interpretation: {'Strong positive' if pearson_corr > 0.7 else 'Moderate positive' if pearson_corr > 0.3 else 'Weak'} linear relationship")

# Calculate Spearman correlation
spearman_corr, spearman_pvalue = spearmanr(data['age'], data['income'])
print(f"\nSpearman Correlation Coefficient: {spearman_corr:.6f}")
print(f"P-Value: {spearman_pvalue:.6f}")
print(f"Interpretation: {'Strong' if abs(spearman_corr) > 0.7 else 'Moderate' if abs(spearman_corr) > 0.3 else 'Weak'} monotonic relationship")

# Difference between Pearson and Spearman
print(f"\nDifference in correlations: {abs(pearson_corr - spearman_corr):.6f}")
print("Pearson measures linear relationships; Spearman measures monotonic relationships")

# Create a larger dataset with multiple variables for correlation matrix
data_extended = pd.DataFrame({
    'age': np.array([23, 25, 30, 35, 40, 28, 32, 38, 45, 50]),
    'income': np.array([35000, 40000, 50000, 60000, 70000, 45000, 55000, 65000, 75000, 85000]),
    'years_experience': np.array([1, 2, 5, 8, 12, 4, 7, 10, 15, 18])
})

print("\nExtended Dataset:")
print(data_extended)

# Compute correlation matrix
corr_matrix = data_extended.corr()
print("\nPearson Correlation Matrix:")
print(corr_matrix)

# Visualize correlations
plt.figure(figsize=(10, 6))
plt.scatter(data['age'], data['income'], s=100, color='blue', alpha=0.7)
z = np.polyfit(data['age'], data['income'], 1)
p = np.poly1d(z)
plt.plot(data['age'], p(data['age']), "r--", linewidth=2, label=f'Linear fit')
plt.xlabel('Age')
plt.ylabel('Income')
plt.title(f'Age vs Income (Pearson r = {pearson_corr:.4f})')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Heatmap for correlation matrix
import seaborn as sns
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, vmin=-1, vmax=1, square=True, cbar_kws={'label': 'Correlation'})
plt.title('Correlation Matrix Heatmap')
plt.tight_layout()
plt.show()
```

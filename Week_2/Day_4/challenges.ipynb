## 1. Data Scoping and Preparation

```python
# Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
from ipywidgets import interact, Dropdown, IntSlider
from IPython.display import display
import warnings
warnings.filterwarnings('ignore')

# Set visualization defaults
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')

print("✓ All libraries imported successfully")
```

### Load and Explore Dataset

```python
# Load the dataset
df = pd.read_csv(r'C:\Users\DELL\Desktop\DI-BOOTCAMP\Week_2\Day_4\Sample - Superstore.csv', encoding='ISO-8859-1')
# Basic data exploration
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\n" + "="*80)
print("DATA TYPES")
print("="*80)
print(df.info())
print("\n" + "="*80)
print("DESCRIPTIVE STATISTICS")
print("="*80)
print(df.describe())
```

### Check for Missing Values and Duplicates

**Approach:**
- Identify missing values and duplicates
- Remove duplicates
- Handle missing values with appropriate strategies (removal or imputation)
- Justify each decision based on the data context

```python
# Check for duplicates
duplicate_count = df.duplicated().sum()
print(f"Duplicate rows found: {duplicate_count}")

if duplicate_count > 0:
    print("Removing duplicate rows...")
    df = df.drop_duplicates()
    print(f"Dataset shape after removing duplicates: {df.shape}")
else:
    print("✓ No duplicate rows found")

# Check for missing values
print("\n" + "="*80)
print("MISSING VALUES ANALYSIS")
print("="*80)
missing_summary = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Count': missing_summary,
    'Missing Percentage': missing_pct
})

print(missing_df[missing_df['Missing Count'] > 0])

if missing_df['Missing Count'].sum() == 0:
    print("\n✓ No missing values found in the dataset")
else:
    print("\nHandling missing values...")
```

### Fix Data Types

Convert date columns to `datetime` objects for proper time-series analysis.

```python
# Convert date columns to datetime
date_columns = ['Order Date', 'Ship Date']

for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col])
        print(f"✓ Converted '{col}' to datetime")

# Verify the conversion
print("\nData types after conversion:")
print(df[date_columns].dtypes)

# Display date range
print(f"\nDate range: {df['Order Date'].min()} to {df['Order Date'].max()}")
```

### Feature Engineering

Create new features to enrich analysis:
- **Profit Margin**: (Profit / Sales) × 100
- **Order Year**: Extract year from Order Date
- **Order Month**: Extract month number
- **Order Month-Year**: Period for aggregation

```python
# Feature engineering
df['Profit Margin'] = (df['Profit'] / df['Sales']) * 100
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month
df['Order Month-Year'] = df['Order Date'].dt.to_period('M')
df['Delivery Time'] = (df['Ship Date'] - df['Order Date']).dt.days

print("✓ New features created:")
print("  - Profit Margin (%)")
print("  - Order Year")
print("  - Order Month")
print("  - Order Month-Year")
print("  - Delivery Time (days)")

# Display sample of new features
print("\nSample of engineered features:")
print(df[['Sales', 'Profit', 'Profit Margin', 'Order Year', 'Order Month', 'Delivery Time']].head(10))

# Summary statistics for new features
print("\n" + "="*80)
print("NEW FEATURES SUMMARY STATISTICS")
print("="*80)
print(df[['Profit Margin', 'Delivery Time']].describe())
```

## 2. Deep-Dive Exploratory Analysis (Matplotlib)

### Time-Series Trend Analysis

Analyzing monthly sales trends to identify patterns such as seasonality, year-over-year changes, and product-specific trends.

```python
# Prepare data for time series analysis
monthly_sales = df.groupby(['Order Month-Year', 'Category'])['Sales'].sum().reset_index()
monthly_sales['Date'] = monthly_sales['Order Month-Year'].dt.to_timestamp()

print("✓ Time series data prepared")
print(f"Date range: {monthly_sales['Date'].min().date()} to {monthly_sales['Date'].max().date()}")
print(f"Number of months: {len(df.groupby('Order Month-Year'))}")
print(f"Categories: {df['Category'].unique().tolist()}")
```

```python
# Interactive time series plot with Matplotlib + ipywidgets
def plot_monthly_sales(category='All'):
    plt.figure(figsize=(14, 7))

    if category == 'All':
        # Plot total sales across all categories
        total_monthly = df.groupby('Order Month-Year')['Sales'].sum()
        plt.plot(total_monthly.index.to_timestamp(), total_monthly.values, 
                marker='o', linewidth=2.5, markersize=5, color='steelblue')
        plt.title('Monthly Sales Trend - All Categories', fontsize=16, fontweight='bold', pad=20)
        plt.fill_between(total_monthly.index.to_timestamp(), total_monthly.values, alpha=0.2, color='steelblue')
    else:
        # Plot sales for specific category
        category_data = monthly_sales[monthly_sales['Category'] == category].sort_values('Date')
        plt.plot(category_data['Date'], category_data['Sales'], 
                marker='o', linewidth=2.5, markersize=5, color='coral', label=category)
        plt.title(f'Monthly Sales Trend - {category}', fontsize=16, fontweight='bold', pad=20)
        plt.fill_between(category_data['Date'], category_data['Sales'], alpha=0.2, color='coral')

    plt.xlabel('Date', fontsize=12, fontweight='bold')
    plt.ylabel('Sales ($)', fontsize=12, fontweight='bold')
    plt.xticks(rotation=45)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# Create interactive widget
categories = ['All'] + sorted(list(df['Category'].unique()))
category_dropdown = Dropdown(options=categories, value='All', description='Category:')
interact(plot_monthly_sales, category=category_dropdown);
```

### Geographic Sales Performance

Identifying high-performing states and analyzing geographic distribution of sales.

```python
# Prepare geographic sales data
state_sales = df.groupby('State')['Sales'].sum().sort_values(ascending=True)

print(f"Total number of states: {len(state_sales)}")
print(f"\nTop 5 states by sales:")
print(state_sales.tail(5))
print(f"\nTop 5 states represent {(state_sales.tail(5).sum()/state_sales.sum()*100):.1f}% of total sales")
```

```python
# Interactive geographic analysis with slider
def plot_top_states(top_n=10):
    plt.figure(figsize=(12, max(6, top_n * 0.4)))

    # Get top N states
    top_states = state_sales.tail(top_n)

    # Create horizontal bar chart with gradient colors
    colors = plt.cm.viridis(np.linspace(0.2, 0.9, len(top_states)))
    bars = plt.barh(range(len(top_states)), top_states.values, color=colors)
    plt.yticks(range(len(top_states)), top_states.index, fontsize=11)
    plt.xlabel('Total Sales ($)', fontsize=12, fontweight='bold')
    plt.ylabel('State', fontsize=12, fontweight='bold')
    plt.title(f'Top {top_n} States by Sales Performance', fontsize=16, fontweight='bold', pad=20)

    # Add value labels on bars
    for i, (state, value) in enumerate(top_states.items()):
        plt.text(value + max(top_states.values()) * 0.01, i, f'${value:,.0f}', 
                va='center', fontsize=10, fontweight='bold')

    plt.grid(axis='x', alpha=0.3)
    plt.tight_layout()
    plt.show()

    print(f"\n📊 Geographic Insights (Top {top_n}):")
    print(f"   • Total states analyzed: {len(state_sales)}")
    print(f"   • Top {top_n} states represent: ${top_states.sum():,.0f} in sales")
    print(f"   • Top {top_n} states concentration: {(top_states.sum()/state_sales.sum())*100:.1f}% of total sales")
    print(f"   • Average sales per top state: ${top_states.mean():,.0f}")

# Create interactive slider
top_n_slider = IntSlider(min=5, max=min(25, len(state_sales)), value=10, description='Top N States:')
interact(plot_top_states, top_n=top_n_slider);
```

### Discount vs. Profit Scatter Plot

Analyzing the relationship between discount rates and profitability across product categories.

```python
# Discount vs Profit Analysis
plt.figure(figsize=(14, 8))

# Create the scatter plot with category colors
sns.scatterplot(data=df, x='Discount', y='Profit', hue='Category', 
               alpha=0.6, s=60, palette='Set2', edgecolor='white', linewidth=0.5)

# Add regression line for overall trend
sns.regplot(data=df, x='Discount', y='Profit', scatter=False, 
           color='red', line_kws={'linewidth': 2.5, 'linestyle': '--'}, label='Overall Trend')

# Customize the plot
plt.title('Discount Strategy Analysis: Impact on Profitability by Category', 
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Discount Rate', fontsize=12, fontweight='bold')
plt.ylabel('Profit ($)', fontsize=12, fontweight='bold')

# Add horizontal line at profit = 0
plt.axhline(y=0, color='black', linestyle='-', alpha=0.4, linewidth=1.5)
plt.text(0.01, 50, 'Break-even line', fontsize=10, alpha=0.7, fontweight='bold')

plt.grid(True, alpha=0.3)
plt.legend(title='Product Category', bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=10)
plt.tight_layout()
plt.show()

# Analytical insights
print("\n💡 DISCOUNT STRATEGY INSIGHTS:")
print("="*80)

high_discount = df[df['Discount'] > 0.2]
print(f"\n📌 High Discount Analysis (>20%):")
print(f"   • Transactions with >20% discount: {len(high_discount):,}")
print(f"   • Percentage of total transactions: {(len(high_discount)/len(df))*100:.1f}%")
print(f"   • Average profit for high discounts: ${high_discount['Profit'].mean():.2f}")
print(f"   • Median profit for high discounts: ${high_discount['Profit'].median():.2f}")

# Loss analysis
high_disc_loss = high_discount[high_discount['Profit'] < 0]
print(f"   • Percentage of high-discount sales with LOSSES: {(len(high_disc_loss)/len(high_discount))*100:.1f}%")

# Category-specific insights
print(f"\n📌 Category-Specific High Discount Impact:")
for category in df['Category'].unique():
    high_disc_cat = df[(df['Discount'] > 0.2) & (df['Category'] == category)]
    if len(high_disc_cat) > 0:
        avg_profit = high_disc_cat['Profit'].mean()
        loss_rate = (high_disc_cat['Profit'] < 0).sum() / len(high_disc_cat) * 100
        print(f"   • {category}: Avg profit at >20% discount = ${avg_profit:.2f} (Loss rate: {loss_rate:.1f}%)")
```

### Multi-Chart Dashboard

```python
# Advanced: Multi-chart dashboard
def create_dashboard():
    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))

    # Chart 1: Monthly sales trend
    monthly_total = df.groupby('Order Month-Year')['Sales'].sum()
    ax1.plot(monthly_total.index.to_timestamp(), monthly_total.values, marker='o', 
            linewidth=2, markersize=4, color='steelblue')
    ax1.fill_between(monthly_total.index.to_timestamp(), monthly_total.values, alpha=0.2, color='steelblue')
    ax1.set_title('Monthly Sales Trend', fontsize=14, fontweight='bold')
    ax1.set_xlabel('Date')
    ax1.set_ylabel('Sales ($)')
    ax1.tick_params(axis='x', rotation=45)
    ax1.grid(True, alpha=0.3)

    # Chart 2: Category performance
    category_sales = df.groupby('Category')['Sales'].sum().sort_values()
    colors = plt.cm.Set3(range(len(category_sales)))
    ax2.bar(category_sales.index, category_sales.values, color=colors, edgecolor='black', linewidth=1.5)
    ax2.set_title('Sales by Category', fontsize=14, fontweight='bold')
    ax2.set_ylabel('Sales ($)')
    for i, v in enumerate(category_sales.values):
        ax2.text(i, v + max(category_sales.values())*0.02, f'${v/1000:.0f}K', 
                ha='center', fontweight='bold')
    ax2.grid(axis='y', alpha=0.3)

    # Chart 3: State performance (top 10)
    top_10_states = state_sales.tail(10)
    colors_states = plt.cm.viridis(np.linspace(0.2, 0.9, len(top_10_states)))
    ax3.barh(range(len(top_10_states)), top_10_states.values, color=colors_states, edgecolor='black', linewidth=1.5)
    ax3.set_yticks(range(len(top_10_states)))
    ax3.set_yticklabels(top_10_states.index)
    ax3.set_title('Top 10 States by Sales', fontsize=14, fontweight='bold')
    ax3.set_xlabel('Sales ($)')
    ax3.grid(axis='x', alpha=0.3)

    # Chart 4: Discount vs Profit
    for category in df['Category'].unique():
        cat_data = df[df['Category'] == category]
        ax4.scatter(cat_data['Discount'], cat_data['Profit'], 
                   label=category, alpha=0.6, s=50, edgecolor='white', linewidth=0.5)
    ax4.set_xlabel('Discount Rate', fontweight='bold')
    ax4.set_ylabel('Profit ($)', fontweight='bold')
    ax4.set_title('Discount vs Profit by Category', fontsize=14, fontweight='bold')
    ax4.legend(loc='best')
    ax4.axhline(y=0, color='black', linestyle='--', alpha=0.5, linewidth=1.5)
    ax4.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()
    
    print("\n✓ Dashboard created successfully")

# Call the dashboard function
create_dashboard()
```

### Outlier Analysis with Annotations

```python
# Advanced: Outlier analysis with annotations
plt.figure(figsize=(14, 9))

# Create base scatter plot
sns.scatterplot(data=df, x='Discount', y='Profit', hue='Category', alpha=0.6, s=60, palette='Set2')

# Identify and annotate outliers
top_3_profitable = df.nlargest(3, 'Profit')
bottom_3_profitable = df.nsmallest(3, 'Profit')

# Annotate top performers
for idx, row in top_3_profitable.iterrows():
    plt.annotate(f'Best: ${row["Profit"]:.0f}', 
                xy=(row['Discount'], row['Profit']),
                xytext=(15, 15), textcoords='offset points',
                bbox=dict(boxstyle='round,pad=0.5', facecolor='green', alpha=0.7, edgecolor='darkgreen', linewidth=1.5),
                arrowprops=dict(arrowstyle='->', connectionstyle='arc3,rad=0.3', color='darkgreen', lw=1.5),
                fontsize=10, fontweight='bold', color='white')

# Annotate worst performers  
for idx, row in bottom_3_profitable.iterrows():
    plt.annotate(f'Worst: ${row["Profit"]:.0f}', 
                xy=(row['Discount'], row['Profit']),
                xytext=(15, -25), textcoords='offset points',
                bbox=dict(boxstyle='round,pad=0.5', facecolor='red', alpha=0.7, edgecolor='darkred', linewidth=1.5),
                arrowprops=dict(arrowstyle='->', connectionstyle='arc3,rad=0.3', color='darkred', lw=1.5),
                fontsize=10, fontweight='bold', color='white')

plt.title('Discount vs Profit Analysis with Outlier Identification', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Discount Rate', fontsize=12, fontweight='bold')
plt.ylabel('Profit ($)', fontsize=12, fontweight='bold')
plt.legend(title='Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\n🔍 OUTLIER INSIGHTS:")
print("\nTop 3 Most Profitable Transactions:")
for i, (idx, row) in enumerate(top_3_profitable.iterrows(), 1):
    print(f"   {i}. {row['Product Name']} - Profit: ${row['Profit']:.2f}, Discount: {row['Discount']:.1%}")

print("\nTop 3 Least Profitable Transactions:")
for i, (idx, row) in enumerate(bottom_3_profitable.iterrows(), 1):
    print(f"   {i}. {row['Product Name']} - Profit: ${row['Profit']:.2f}, Discount: {row['Discount']:.1%}")
```

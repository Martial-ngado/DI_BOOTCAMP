# Daily Challenge: Interactive Data Visualization with Matplotlib and Seaborn

This notebook downloads the US Superstore dataset, performs basic cleaning, and produces interactive visualizations using Matplotlib (with `mplcursors`) and Plotly, plus elegant static charts with Seaborn. Explanations and insights are included in markdown cells.

```python
# Install any missing packages (runs in the notebook environment).
import sys, subprocess
subprocess.check_call([sys.executable, '-m', 'pip', 'install', 'mplcursors', 'plotly', 'ipympl'])
```

```python
# Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import mplcursors
import plotly.express as px
import seaborn as sns
sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10,5)
```

```python
# Load the US Superstore dataset directly from the provided URL
url = 'https://github.com/devtlv/Datasets-DA-Bootcamp-2-/raw/refs/heads/main/Week%205%20-%20Data%20Processing/W5D5%20-%20Mini-project%20-%20bis/US%20Superstore%20data.xls'
df = pd.read_excel(url)
# Quick peek
display(df.head())
# Basic cleaning and preprocessing
# Ensure date column is datetime and numeric columns are numeric
if 'Order Date' in df.columns:
    df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
if 'Sales' in df.columns:
    df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
if 'Profit' in df.columns:
    df['Profit'] = pd.to_numeric(df['Profit'], errors='coerce')
if 'Discount' in df.columns:
    df['Discount'] = pd.to_numeric(df['Discount'], errors='coerce')
# Drop rows missing essential fields
df = df.dropna(subset=['Order Date','Sales'])
# Add helper columns
df['Year'] = df['Order Date'].dt.year
print('Rows after cleaning:', len(df))
```

## Matplotlib: Interactive Line Chart (Sales Trends by Year)
The chart below uses Matplotlib for plotting and `mplcursors` to make data-point tooltips interactive in notebook frontends that support it.

```python
# Aggregate sales by year
sales_by_year = df.groupby('Year', as_index=False)['Sales'].sum().sort_values('Year')
# Plot with Matplotlib
fig, ax = plt.subplots()
ax.plot(sales_by_year['Year'], sales_by_year['Sales'], marker='o', linestyle='-')
ax.set_title('Total Sales by Year')
ax.set_xlabel('Year')
ax.set_ylabel('Sales (USD)')
ax.grid(True)
# Enable interactive annotations on hover (mplcursors)
mplcursors.cursor(ax.lines, hover=True)
plt.show()
# Observations can be added below as comments or markdown
```

## Matplotlib: Interactive Map of Sales by Country
Here we use Plotly (`px.choropleth`) for an interactive world map because Plotly offers robust, browser-based interactivity for geographic data. If you specifically need a Matplotlib-based map (e.g., with GeoPandas), I can provide that alternative, but it typically requires additional heavy dependencies.

# NumPy Exercises
Solutions and outputs for Exercises 1–10.

```python
import numpy as np

# Exercise 1
a1 = np.arange(10)
print("Exercise 1 output:")
print(a1)

# Exercise 2
a2 = np.array([3.14, 2.17, 0, 1, 2]).astype(int)
print("\nExercise 2 output:")
print(a2)

# Exercise 3
a3 = np.arange(1, 10).reshape(3, 3)
print("\nExercise 3 output:")
print(a3)

# Exercise 4 (reproducible random)
np.random.seed(0)
a4 = np.random.rand(4, 5)
print("\nExercise 4 output:")
print(np.round(a4, 2))

# Exercise 5
arr5 = np.array([[21,22,23,22,22],[20,21,22,23,24],[21,22,23,22,22]])
print("\nExercise 5 array:")
print(arr5)
print("Second row:")
print(arr5[1])

# Exercise 6
print("\nExercise 6 output:")
print(a1[::-1])

# Exercise 7
print("\nExercise 7 output:")
print(np.eye(4))

# Exercise 8
s = a1.sum()
m = a1.mean()
print("\nExercise 8 output:")
print(f"Sum: {s}, Average: {m}")

# Exercise 9
a9 = np.arange(1, 21).reshape(4, 5)
print("\nExercise 9 output:")
print(a9)

# Exercise 10
arr10 = np.arange(1, 10)
print("\nExercise 10 output:")
print(arr10[arr10 % 2 == 1])
```

# Daily Challenge: NumPy, Pandas, and Matplotlib Integration

## Objective
Analyze temperature trends across different regions using NumPy, Pandas, and Matplotlib.

## Tasks
1. Data Preparation
2. Data Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Set random seed for reproducibility
np.random.seed(42)
```

## Task 1: Data Preparation

Generate synthetic temperature data using NumPy and create a Pandas DataFrame with:
- 10 cities as index
- 12 months as columns
- Temperature values ranging from -5 to 35 degrees Celsius

```python
# Define city names and months
cities = ['New York', 'London', 'Tokyo', 'Sydney', 'Paris', 'Dubai', 'Berlin', 'Toronto', 'Mumbai', 'São Paulo']
months = ['January', 'February', 'March', 'April', 'May', 'June', 
          'July', 'August', 'September', 'October', 'November', 'December']

# Generate synthetic temperature data using NumPy
# Temperature range: -5 to 35 degrees Celsius
# Shape: (10 cities, 12 months)
temperature_data = np.random.uniform(low=-5, high=35, size=(10, 12))

# Create a Pandas DataFrame
df_temperatures = pd.DataFrame(
    data=temperature_data,
    index=cities,
    columns=months
)

print("Temperature DataFrame:")
print(df_temperatures)
print("\nDataFrame shape:", df_temperatures.shape)
print("\nFirst few rows:")
print(df_temperatures.head())
```

## Task 2: Data Analysis

Calculate annual average temperatures and identify cities with extreme temperatures.

```python
# Calculate annual average temperature for each city
annual_avg_temp = df_temperatures.mean(axis=1)

print("Annual Average Temperature for Each City:")
print(annual_avg_temp)
print("\n" + "="*60 + "\n")

# Find the city with the highest average temperature
city_highest = annual_avg_temp.idxmax()
temp_highest = annual_avg_temp.max()

print(f"City with HIGHEST average temperature:")
print(f"  City: {city_highest}")
print(f"  Temperature: {temp_highest:.2f}°C")
print("\n" + "="*60 + "\n")

# Find the city with the lowest average temperature
city_lowest = annual_avg_temp.idxmin()
temp_lowest = annual_avg_temp.min()

print(f"City with LOWEST average temperature:")
print(f"  City: {city_lowest}")
print(f"  Temperature: {temp_lowest:.2f}°C")
print("\n" + "="*60 + "\n")

# Additional statistics
print("Summary Statistics:")
print(f"  Overall Mean Temperature: {annual_avg_temp.mean():.2f}°C")
print(f"  Overall Std Deviation: {annual_avg_temp.std():.2f}°C")
print(f"  Temperature Range: {temp_lowest:.2f}°C to {temp_highest:.2f}°C")
```

## Visualization: Temperature Trends

Create visualizations to better understand temperature patterns.

```python
# Create a figure with multiple subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Bar chart of annual average temperatures by city
ax1 = axes[0, 0]
annual_avg_temp.sort_values(ascending=False).plot(kind='bar', ax=ax1, color='skyblue', edgecolor='navy')
ax1.set_title('Annual Average Temperature by City', fontsize=12, fontweight='bold')
ax1.set_xlabel('City')
ax1.set_ylabel('Temperature (°C)')
ax1.grid(axis='y', alpha=0.3)
ax1.tick_params(axis='x', rotation=45)

# 2. Line plot showing temperature trends across months for all cities
ax2 = axes[0, 1]
for city in cities:
    ax2.plot(months, df_temperatures.loc[city], marker='o', label=city, alpha=0.7)
ax2.set_title('Monthly Temperature Trends by City', fontsize=12, fontweight='bold')
ax2.set_xlabel('Month')
ax2.set_ylabel('Temperature (°C)')
ax2.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=8)
ax2.grid(alpha=0.3)
ax2.tick_params(axis='x', rotation=45)

# 3. Box plot showing temperature distribution
ax3 = axes[1, 0]
df_temperatures.T.plot(kind='box', ax=ax3)
ax3.set_title('Temperature Distribution by City', fontsize=12, fontweight='bold')
ax3.set_ylabel('Temperature (°C)')
ax3.grid(axis='y', alpha=0.3)
ax3.tick_params(axis='x', rotation=45)

# 4. Heatmap using imshow (simulating a heatmap)
ax4 = axes[1, 1]
im = ax4.imshow(df_temperatures.values, cmap='RdYlBu_r', aspect='auto')
ax4.set_title('Temperature Heatmap (Cities × Months)', fontsize=12, fontweight='bold')
ax4.set_xlabel('Month')
ax4.set_ylabel('City')
ax4.set_xticks(np.arange(len(months)))
ax4.set_yticks(np.arange(len(cities)))
ax4.set_xticklabels(months, rotation=45, ha='right')
ax4.set_yticklabels(cities)
cbar = plt.colorbar(im, ax=ax4)
cbar.set_label('Temperature (°C)', rotation=270, labelpad=15)

plt.tight_layout()
plt.show()

print("Visualization complete!")
```
